```python
# STEP 2: Voice Reference
import os, glob, subprocess

clean_ref = "/kaggle/working/clean_voice_ref.wav"
master_refs = glob.glob("/kaggle/input/**/shravan_sir_master_ref.wav", recursive=True) + glob.glob("/kaggle/input/**/intro_0_20.wav", recursive=True)
if master_refs:
    raw_wav = master_refs[0]
    print(f"✅ Found existing dataset voice sample: {raw_wav}")
else:
    print("⏳ Downloading clean voice reference from YouTube...")
    os.makedirs("/kaggle/working/ref", exist_ok=True)
    raw_wav = "/kaggle/working/ref/intro_0_20.wav"
    subprocess.run([
        "yt-dlp", "-f", "ba", "--download-sections", "*0:00-0:20",
        "-x", "--audio-format", "wav",
        "-o", "/kaggle/working/ref/intro_0_20.%(ext)s",
        "https://youtu.be/ixCAZT2vHcM"
    ], check=True)

subprocess.run(["ffmpeg", "-y", "-i", raw_wav, "-ac", "1", "-ar", "24000", "-af", "highpass=f=80,loudnorm=I=-16:TP=-1.5:LRA=11", clean_ref], check=True)
print(f"🎉 Clean Voice Reference Ready: {clean_ref}")
```

```python
# STEP 3: Load Chatterbox Multilingual Model
import torch
torch.backends.cudnn.enabled = False
import torchaudio as ta
from chatterbox.mtl_tts import ChatterboxMultilingualTTS

model = ChatterboxMultilingualTTS.from_pretrained(device="cuda")
print("✅ Chatterbox Multilingual Model Ready!")
```

```python
# STEP 4: Generate 8 Fixed Audio Clips & Zip
import os, shutil, glob, torchaudio as ta
from IPython.display import Audio, display

FIX_SCENES = [
    {
        "id": "speech_ans_006",
        "text": "अरे कुछ लोग अभी भी सो रहे हैं। जल्दी से चैट में एक्टिव हो जाओ, लाइव टेस्ट शुरू हो चुका है। बिल्कुल सही। सार्वत्रिक गुरुत्वाकर्षण स्थिरांक का विमीय सूत्र सीधे याद रखिए, एम की घात माइनस 1, एल क्यूब, टी की घात माइनस 2। सही उत्तर है विकल्प सी, एम की घात माइनस 1, एल क्यूब, टी की घात माइनस 2।"
    },
    {
        "id": "speech_q_033",
        "text": "33 नंबर सवाल कक्षा 10 का है, बहुत शानदार सवाल, विद्युत प्रतिरोध का मात्रक क्या होता है? देखते हैं किसका जवाब सबसे पहले आता है, चैट में लिखिए।"
    },
    {
        "id": "speech_ans_039",
        "text": "अरे चैट की स्पीड देखकर लग रहा है कि आज सब लोग पूरी तैयारी के साथ आए हैं। सही उत्तर है विकल्प ए, 20 हर्ट्ज से 20000 हर्ट्ज।"
    },
    {
        "id": "speech_ans_079",
        "text": "अरे राहुल भाई, अबकी बार सही आंसर आया है, बहुत बढ़िया सुधार। बिल्कुल सही, दही में लैक्टिक अम्ल बन जाता है, जिसे पुनः दूध में नहीं बदला जा सकता। सही उत्तर है विकल्प बी, रासायनिक परिवर्तन।"
    },
    {
        "id": "speech_ans_094",
        "text": "अरे रोहित भाई, चैट में गलत आंसर देकर डिलीट मत करो, हम सब देख रहे हैं। शानदार, लगातार सही आंसर आ रहे हैं। बिल्कुल सही, लोटनिक घर्षण का मान सबसे कम होता है। इसी कारण मशीनों में पहिए और बॉल बेयरिंग लगाकर घर्षण घटाते हैं। सही उत्तर है विकल्प सी, लोटनिक घर्षण।"
    },
    {
        "id": "speech_q_103",
        "text": "स्क्रीन पर अगला सवाल, 103 नंबर, कमरे के ताप पर द्रव अवस्था में पाई जाने वाली एकमात्र अधातु कौन सी है? चैट बॉक्स में फटाफट अपना उत्तर लिखिए।"
    },
    {
        "id": "speech_ans_127",
        "text": "अरे चैट में कुछ लोग तुक्का मार रहे हैं। तुक्के से माइनस एक अंक मिलेगा भाई। बिल्कुल सही, बर्फ पिघलकर पानी बनती है और पानी को पुनः जमाकर बर्फ बनाया जा सकता है, अतः यह उत्क्रमणीय परिवर्तन है। सही उत्तर है विकल्प बी, बर्फ का पिघलना।"
    },
    {
        "id": "speech_ans_192",
        "text": "अरे भाई थोड़ा सा दिमाग लगाओ, अंतिम पलों में गलती मत करो। बिल्कुल सही, खरपतवार फसल से पोषक तत्व, जल और धूप छीन लेते हैं। इन्हें खरपतवारनाशी टू फोर डी से नष्ट करते हैं। सही उत्तर है विकल्प डी, खरपतवार।"
    }
]

OUT_DIR = "/kaggle/working/fixed_8_audios"
os.makedirs(OUT_DIR, exist_ok=True)
voice_ref = "/kaggle/input/mastervoice/shravan_sir_master_ref.wav" if os.path.exists("/kaggle/input/mastervoice/shravan_sir_master_ref.wav") else "/kaggle/working/clean_voice_ref.wav"

print(f"🎙️ Using Voice Reference: {voice_ref}")
print("🚀 Generating 8 Fixed Audio Clips (~45 seconds)...\n")

for idx, s in enumerate(FIX_SCENES, 1):
    out_path = os.path.join(OUT_DIR, f"{s['id']}.wav")
    wav = model.generate(
        text=s["text"],
        language_id="hi",
        audio_prompt_path=voice_ref,
        exaggeration=0.45,
        cfg_weight=0.55,
        temperature=0.45,
        repetition_penalty=2.0
    )
    ta.save(out_path, wav, model.sr)
    print(f"✅ [{idx}/8] Done: {s['id']}.wav")
    display(Audio(out_path))

zip_path = shutil.make_archive('/kaggle/working/fixed_8_audios', 'zip', OUT_DIR)
print(f"\n🎉 ALL 8 CLIPS FIXED! Download zip from Output: {zip_path}")
```
